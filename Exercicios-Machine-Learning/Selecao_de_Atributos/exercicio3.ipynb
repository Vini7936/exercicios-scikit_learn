```python
import matplotlib.pyplot as plt 
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFECV
from sklearn.model_selection import StratifiedKFold
```

```python
X, y = make_classification(n_samples = 500, n_features = 30, n_informative = 5, n_redundant = 10, random_state = 42)
```

```python
modelo = LogisticRegression(max_iter = 1000)
cv = StratifiedKFold(n_splits = 5)
```

```python
rfecv = RFECV(estimator = modelo, step = 1, cv=cv, scoring = 'accuracy')
rfecv.fit(X, y)
```

```python
print(f"Número ideal de atributos: {rfecv.n_features_}")
print(f"Atributos selecionados (máscara): {rfecv.support_}")
```

```python
plt.figure(figsize=(10, 6))
plt.xlabel("Número de Atributos Selecionados")
plt.ylabel("Acurácia da Validação Cruzada (com σ)")
plt.plot(range(1, len(rfecv.cv_results_['mean_test_score']) + 1), 
         rfecv.cv_results_['mean_test_score'], marker='o')
plt.title("RFECV - Desempenho do Modelo por Quantidade de Atributos")
plt.grid(True)
plt.show()
```
