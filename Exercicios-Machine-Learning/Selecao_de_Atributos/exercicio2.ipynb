```python
from sklearn.datasets import load_digits 
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
dados = load_digits()
X, y = dados.data, dados.target
```

```python
# Verificar a quantidade de atributos
print(f"Número original de atributos: {X.shape[1]}")
```

```python
seletor = SelectKBest(score_func = f_classif, k=15)
X_novo = seletor.fit_transform(X, y)
```

```python
mascara = seletor.get_support()
```

```python
indices_escolhidos = [i for i, ativo in enumerate(mascara) if ativo]
```

```python
print(f"Número reduzido de atributos: {X_novo.shape[1]}")
print(f"Índices dos 15 melhores pixels/atributos: {indices_escolhidos}")
```
