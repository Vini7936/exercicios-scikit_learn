```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, AgglomerativeClustering
```

```python
X_base, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.5, random_state=42)
outliers = np.random.uniform(low=-10.0, high=10.0, size=(20, 2))
X = np.vstack([X_base, outliers])
```

```python
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X)
hierarchical = AgglomerativeClustering(n_clusters=3, linkage='ward').fit_predict(X)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
ax1.scatter(X[:, 0], X[:, 1], c=kmeans, cmap='viridis', s=40, alpha=0.8)
ax1.set_title("K-Means (Sensível a Outliers)")
ax1.grid(True)

ax2.scatter(X[:, 0], X[:, 1], c=hierarchical, cmap='viridis', s=40, alpha=0.8)
ax2.set_title("Agrupamento Hierárquico (Ward)")
ax2.grid(True)

plt.tight_layout()
plt.show()
```
