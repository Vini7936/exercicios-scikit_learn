```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth, association_rules
```

```python
transacoes_complexas = [
    ['Arroz', 'Feijão', 'Óleo'],
    ['Arroz', 'Feijão', 'Carne', 'Cerveja'],
    ['Feijão', 'Carne', 'Cerveja'],
    ['Arroz', 'Óleo'],
    ['Arroz', 'Feijão', 'Carne'],
    ['Arroz', 'Feijão', 'Cerveja', 'Carne'],
    ['Óleo', 'Arroz'],
    ['Carne', 'Cerveja']
]
```

```python
te = TransactionEncoder()
dados = pd.DataFrame(te.fit(transacoes_complexas).transform(transacoes_complexas), columns=te.columns_)
```

```python
itemsets_frequent = fpgrowth(dados, min_support=0.25, use_colnames=True)
```

```python
regras_totais = association_rules(itemsets_frequent, metric="confidence", min_threshold=0.5)
```

```python
regras_filtradas = regras_totais[
    (regras_totais['lift'] > 1.2) & 
    (regras_totais['confidence'] > 0.7)
].sort_values(by='lift', ascending=False)

print(f"Total de regras geradas inicialmente: {len(regras_totais)}")
print(f"Total de regras após filtragem restritiva: {len(regras_filtradas)}\n")
print(regras_filtradas[['antecedents', 'consequents', 'support', 'confidence', 'lift']].to_string(index=False))
```
