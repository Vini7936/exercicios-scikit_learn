```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
```

```python
transacoes = [
    ['Leite', 'Pão', 'Manteiga'],
    ['Pão', 'Manteiga'],
    ['Leite', 'Pão'],
    ['Leite', 'Café'],
    ['Leite', 'Pão', 'Manteiga', 'Café']
]
```

```python
te = TransactionEncoder()
te_ary = te.fit(transacoes).transform(transacoes)
dados = pd.DataFrame(te_ary, columns = te.columns_)
```

```python
print("Tabela One-Hot das Transações:")
print(dados)
print("\n" + "="*40 + "\n")
```

```python
frequent_itemsets = apriori(dados, min_support=0.4, use_colnames=True)
```

```python
print("Conjuntos de Itens Frequentes:")
print(frequent_itemsets)
```
