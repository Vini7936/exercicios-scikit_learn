```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
X, _ = make_blobs(n_samples = 500, centers = 4, cluster_std = 0.7, random_state = 42)
```

```python
inercias = []
silhuetas = []
k_valores = range(2, 11)
```

```python
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X)
    inercias.append(km.inertia_)
    if k >= 2:
        silhuetas.append(silhouette_score(X, km.labels_))
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.plot(range(1, 11), inercias, marker='o', color='b')
ax1.set_title('Método do Cotovelo (Inércia)')
ax1.set_xlabel('Número de clusters (k)')
ax1.set_ylabel('Inércia')
ax1.grid(True)

ax2.plot(k_valores, silhuetas, marker='s', color='g')
ax2.set_title('Silhouette Score')
ax2.set_xlabel('Número de clusters (k)')
ax2.set_ylabel('Score Médio')
ax2.grid(True)

plt.tight_layout()
plt.show()
```
